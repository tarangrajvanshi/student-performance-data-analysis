# Week 2 – Student Performance Preliminary Analysis

This notebook documents data acquisition, cleaning, descriptive analysis, and preliminary EDA using the UCI Student Performance Mathematics dataset.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

## 1. Data Acquisition

Option A: download `student-mat.csv` from the official UCI page and place it in `data/raw/`.

Option B: use the UCI Python package:


```python
from ucimlrepo import fetch_ucirepo

student_performance = fetch_ucirepo(id=320)
X = student_performance.data.features
y = student_performance.data.targets
df = pd.concat([X, y], axis=1)
df.head()
```

## 2. Initial Inspection


```python
print('Shape:', df.shape)
display(df.head())
df.info()
```

```python
print('Missing values:')
display(df.isna().sum())
print('Duplicate rows:', df.duplicated().sum())
```

## 3. Cleaning and Validation


```python
clean_df = df.copy()

for col in clean_df.select_dtypes(include='object').columns:
    clean_df[col] = clean_df[col].astype(str).str.strip().str.lower()

clean_df = clean_df.drop_duplicates().reset_index(drop=True)

invalid_g3 = clean_df[(clean_df['G3'] < 0) | (clean_df['G3'] > 20)]
print('Invalid G3 rows:', len(invalid_g3))
```

## 4. Descriptive Statistics


```python
display(clean_df.describe())
print('G3 mean:', clean_df['G3'].mean())
print('G3 median:', clean_df['G3'].median())
```

## 5. Preliminary Visualizations


```python
sns.histplot(clean_df['G3'], bins=11, kde=True)
plt.title('Distribution of Final Grades')
plt.xlabel('Final Grade (G3)')
plt.show()
```

```python
sns.boxplot(data=clean_df, x='studytime', y='G3')
plt.title('Final Grade by Study Time')
plt.show()
```

```python
sns.scatterplot(data=clean_df, x='absences', y='G3')
plt.title('Absences vs Final Grade')
plt.show()
```

```python
cols = ['studytime', 'failures', 'absences', 'G1', 'G2', 'G3']
sns.heatmap(clean_df[cols].corr(), annot=True, cmap='coolwarm')
plt.title('Selected Variable Correlations')
plt.show()
```

## 6. Preliminary Interpretation

- Examine the final-grade distribution and identify possible skew or unusual values.
- Compare final grades across study-time categories.
- Investigate the relationship between absences and final grade.
- Compare earlier grades (`G1`, `G2`) with the final grade (`G3`).
- Treat correlations as associations rather than causal evidence.
